```python
"""
[2주차] pandas DataFrame, indexing, filtering, dtype
"""
import pandas as pd
import numpy as np
```

```python

# 실습용 무역 데이터 생성
np.random.seed(42)
# seed(42)가 있으면 다른 사람이 똑같은 코드를 실행해도 같은 랜덤 결과를 얻을 수 있어.

df = pd.DataFrame({
    "country": ["Korea", "Japan", "China", "USA", "Germany"] * 3,
    "year": np.repeat([2022, 2023, 2024], 5),
    "export": np.random.randint(100, 1000, 15).astype(float),
    "import_": np.random.randint(100, 1000, 15).astype(float),
    "region": (["Asia", "Asia", "Asia", "America", "Europe"] * 3),
})
```

```python
df.head()
# df.tail()
# df.info()
```

```python

# ------------------------------------------------------------
# 1. 데이터 구조 파악: 분석의 첫 습관
# ------------------------------------------------------------
print(df.head()) # 앞 5개
print("\nshape:", df.shape)
print("\ndtypes:\n", df.dtypes)
print("\ninfo:"); df.info()
```

```python
# ------------------------------------------------------------
# 2. Indexing: loc(라벨) vs iloc(위치)
# ------------------------------------------------------------
print("\nloc  예시:\n", df.loc[0:2, ["country", "export"]])
print("\niloc 예시:\n", df.iloc[0:4, 0:4])
```

```python
df['year']
# df["export"]
```

```python
# ------------------------------------------------------------
# 3. Filtering (Boolean indexing)
# ------------------------------------------------------------
cond = (df["year"] == 2024) & (df["export"] > 400)
print("\n2024년 수출 400 초과:\n", df[cond])

# query() 방식도 함께
print("\nquery 방식:\n", df.query("year == 2024 and export > 400"))
```

```python
print("\n cond 결과값 \n", cond)
```

```python
# ------------------------------------------------------------
# 4. 파생변수 생성과 dtype 관리
# ------------------------------------------------------------
print("변경전 \n", df.dtypes)
df["trade_balance"] = df["export"] - df["import_"]
df["region"] = df["region"].astype("category")   # 메모리 절약 + 범주형 명시
df["year"] = df["year"].astype("int16")
print("\ndtype 변경 후:\n", df.dtypes)
```

```python
# ------------------------------------------------------------
# 5. groupby 기초
# ------------------------------------------------------------
print("\n국가별 평균 수출:\n", df.groupby("country", observed=True)["export"].mean())
print("\n지역-연도별 수출 합계:\n",
      df.groupby(["region", "year"])["export"].sum())
```

```python
# 실습 TASK
# 1. 컬럼명을 적절하게 변환(AI활용)
# 2. Date 컬럼 타입을 object에서 date 타입으로 변환(AI활용)
# 3. Rented_Bike_Count 컬럼을 시계열데이터로해서 date 즉 time별로 그래프 그리기
```

### 1. 컬럼명 변환 (Column Renaming)

데이터프레임의 컬럼명을 분석하기 쉽고 명확한 영어 이름으로 변경합니다. 특수문자를 제거하고, 단위 정보를 포함하여 컬럼의 의미를 명확히 합니다.

```python
df_2.columns = [
    'Date',
    'Rented_Bike_Count',
    'Hour',
    'Temperature_C',
    'Humidity_percent',
    'Wind_speed_ms',
    'Visibility_10m',
    'Dew_point_temperature_C',
    'Solar_Radiation_MJ_m2',
    'Rainfall_mm',
    'Snowfall_cm',
    'Seasons',
    'Holiday',
    'Functioning_Day'
]
df_2.info()
```

### 2. 'Date' 컬럼 타입 변환 및 시간 데이터 생성

'Date' 컬럼을 `object` 타입에서 `datetime` 타입으로 변환하고, 'Date'와 'Hour' 컬럼을 합쳐서 완벽한 시계열 데이터를 위한 `datetime` 인덱스를 생성합니다. `errors='coerce'`를 사용하여 변환할 수 없는 값은 `NaT` (Not a Time)로 처리합니다.

```python
df_2['Date'] = pd.to_datetime(df_2['Date'], format='%d/%m/%Y', errors='coerce')
# 'Date'와 'Hour'를 결합하여 완전한 datetime 객체 생성
df_2['Datetime'] = df_2['Date'] + pd.to_timedelta(df_2['Hour'], unit='h')
df_2.set_index('Datetime', inplace=True)
df_2.info()
```

### 3. 'Rented_Bike_Count' 시계열 그래프 그리기

이제 'Rented_Bike_Count' 컬럼을 시계열 데이터로 사용하여 시간에 따른 자전거 대여량 변화를 그래프로 시각화합니다. 이를 통해 시간별, 일별, 계절별 패턴을 파악할 수 있습니다.

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(15, 7))
sns.lineplot(data=df_2, x=df_2.index, y='Rented_Bike_Count')
plt.title('Rented Bike Count Over Time')
plt.xlabel('Date and Time')
plt.ylabel('Rented Bike Count')
plt.grid(True)
plt.tight_layout()
plt.show()
```
