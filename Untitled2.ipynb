```python

import pandas as pd

base = "https://wwwn.cdc.gov/Nchs/Data/Nhanes/Public/2017/DataFiles/"

# 1. 실제 NHANES 자료 불러오기
diq = pd.read_sas(base + "DIQ_J.XPT", format="xport")
glu = pd.read_sas(base + "GLU_J.XPT", format="xport")
ghb = pd.read_sas(base + "GHB_J.XPT", format="xport")

# 2. SEQN을 기준으로 조인
df = diq.merge(
    glu[["SEQN", "LBXGLU", "WTSAF2YR"]],
    on="SEQN",
    how="left",
    validate="one_to_one"
)

df = df.merge(
    ghb[["SEQN", "LBXGH"]],
    on="SEQN",
    how="left",
    validate="one_to_one"
)

# 3. 조인 결과 확인
print("최종 데이터 크기:", df.shape)
print("\n주요 변수 결측치:")
print(df[["DIQ010", "LBXGLU", "LBXGH"]].isna().sum())

display(df.head())

# 4. CSV로 저장
df.to_csv("NHANES_diabetes_joined.csv", index=False)
```
