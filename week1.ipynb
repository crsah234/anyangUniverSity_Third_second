```python
# import 의미
##
import time
import numpy as np
import pandas as pd
```

```python
# ------------------------------------------------------------
# 1. Loop vs Vectorization : 실행시간 직접 비교
# ------------------------------------------------------------
data = np.random.rand(5000000)

# (1) Python loop 방식
start = time.time()
result_loop = []
for x in data:
    result_loop.append(x * 2)
t_loop = time.time() - start

# (2) Vectorized 방식
start = time.time()
result_vec = data * 2
t_vec = time.time() - start

print(f"loop 방식      : {t_loop:.3f}초")
print(f"vectorized 방식: {t_vec:.4f}초")
print(f"속도 차이      : 약 {t_loop / t_vec:.0f}배")

## 차이나는 이유
# result_loop는 직접 꺼내와서 넣는 거고
# result_vec 은 즉, 지금 배우고 있는 벡터화(vectorization)에서는 data가 NumPy 배열이면
# 처럼 반복문 없이 배열 전체에 연산을 적용할 수 있다는 게 핵심이야.

```

```python
# ------------------------------------------------------------
# 2. Broadcasting 맛보기
# ------------------------------------------------------------
gdp = np.array([100, 200, 300])        # 단위: 십억 달러
growth = 1.03
print("\n내년 GDP 예상:", gdp * growth)  # 스칼라가 배열 전체에 적용됨
```

```python
# ------------------------------------------------------------
# 3. Tidy Data 개념: "한 행 = 한 관측치, 한 열 = 한 변수"
# ------------------------------------------------------------
# 나쁜 예 (변수가 열 이름에 숨어 있음)
messy = pd.DataFrame({
    "country": ["Korea", "Japan"],
    "GDP_2022": [100, 200],
    "GDP_2023": [105, 205],
})
print("\n[messy data]"); print(messy)

# 좋은 예 (tidy)

# https://koreadatascientist.tistory.com/12
# melt 함수 알려줌
tidy = messy.melt(id_vars="country", var_name="year", value_name="GDP")
tidy["year"] = tidy["year"].str.replace("GDP_", "").astype(int)
print("\n[tidy data]"); print(tidy)
```
